## Task 2: Calculating Weekly Financial Totals and Averages (`week_totals()`)
**Approach:**  
We build a reusable function named `week_totals()` that accepts the list of daily records and aggregates total items sold, total revenue, total operating expenses, and net profit. It also computes daily averages and the overall net profit margin percentage. We format the docstring rigorously to outline parameter expectations and returned key-value dictionary outputs.

```python
def week_totals(data):
    """
    Calculate aggregate financial totals, net profit, and daily averages for the week.

    Parameters:
        data (list of dict): List containing daily business transaction records.
            Each dictionary must include 'items_sold', 'revenue', and 'expenses'.

    Returns:
        dict: Comprehensive summary dictionary containing:
            - total_items_sold (int): Aggregate garments sold across all recorded days.
            - total_revenue (int/float): Total gross earnings in Kenyan Shillings (KES).
            - total_expenses (int/float): Total operating expenses incurred in KES.
            - total_profit (int/float): Net earnings (revenue - expenses) in KES.
            - avg_daily_revenue (float): Mean daily revenue earned.
            - avg_daily_expenses (float): Mean daily operational expense.
            - avg_daily_profit (float): Mean daily net profit.
            - avg_items_sold (float): Mean units sold per day.
            - overall_profit_margin (float): Net profit as a percentage of gross revenue.
    """
    if not data:
        return {}

    num_days = len(data)
    total_items = sum(record["items_sold"] for record in data)
    total_rev = sum(record["revenue"] for record in data)
    total_exp = sum(record["expenses"] for record in data)
    net_profit = total_rev - total_exp

    margin = (net_profit / total_rev * 100) if total_rev > 0 else 0.0

    return {
        "total_items_sold": total_items,
        "total_revenue": total_rev,
        "total_expenses": total_exp,
        "total_profit": net_profit,
        "avg_daily_revenue": total_rev / num_days,
        "avg_daily_expenses": total_exp / num_days,
        "avg_daily_profit": net_profit / num_days,
        "avg_items_sold": total_items / num_days,
        "overall_profit_margin": margin
    }

# Execute and display week_totals
summary = week_totals(stall_data)
print("--- Wanjiku's Weekly Performance Summary ---")
print(f"Total Items Sold:       {summary['total_items_sold']} units")
print(f"Total Gross Revenue:    KES {summary['total_revenue']:,}")
print(f"Total Operating Costs:  KES {summary['total_expenses']:,}")
print(f"Total Net Profit:       KES {summary['total_profit']:,}")
print(f"Overall Profit Margin:  {summary['overall_profit_margin']:.2f}%")
print(f"Daily Average Revenue:  KES {summary['avg_daily_revenue']:,.2f}")
print(f"Daily Average Profit:   KES {summary['avg_daily_profit']:,.2f}")
print(f"Average Units / Day:    {summary['avg_items_sold']:.1f} units")
```

## Task 3: Daily Revenue and Profitability Classification (`classify_day()`)
**Approach:**  
Gikomba Market dynamics vary sharply based on bale opening cycles, wholesale foot-traffic, and weekday weather. We implement the `classify_day()` function with clear threshold logic to categorize each day into qualitative operational tiers ('Peak Bale Day', 'High Performer', 'Moderate Day', or 'Slow / Clearance Day'). The function also determines if the day meets Wanjiku's target net profit margin threshold of 50%.

```python
def classify_day(revenue, expenses, benchmark=25000):
    """
    Classify an individual business day into operational tiers based on revenue and profitability.

    Parameters:
        revenue (int/float): Gross daily sales earned in KES.
        expenses (int/float): Total operating costs incurred in KES.
        benchmark (int/float, optional): Mid-level revenue threshold benchmark in KES. Defaults to 25,000.

    Returns:
        dict: Evaluation containing:
            - category (str): Performance tier ('Peak Bale Day', 'High Performer', 'Moderate Day', 'Slow / Clearance Day').
            - profit (int/float): Daily net profit in KES.
            - margin (float): Daily net profit margin percentage.
            - is_healthy_margin (bool): True if net profit margin exceeds the 50% target threshold.
    """
    profit = revenue - expenses
    margin = (profit / revenue * 100) if revenue > 0 else 0.0
    is_healthy = margin >= 50.0

    if revenue >= 40000:
        tier = "Peak Bale Day"
    elif revenue >= benchmark:
        tier = "High Performer"
    elif revenue >= 15000:
        tier = "Moderate Day"
    else:
        tier = "Slow / Clearance Day"

    return {
        "category": tier,
        "profit": profit,
        "margin": margin,
        "is_healthy_margin": is_healthy
    }

# Test classify_day across all days
print("Daily Performance Classification:")
print("-" * 80)
for record in stall_data:
    res = classify_day(record["revenue"], record["expenses"])
    print(f"{record['day']:<10} | Revenue: KES {record['revenue']:,} | Profit: KES {res['profit']:>6,} ({res['margin']:.1f}%) | Tier: {res['category']}")
```

## Task 4: Inventory Depletion & Bale Reorder Management (`check_reorder_status()`)
**Approach:**  
Stockouts on fast-moving mitumba items directly translate to lost revenue for Wanjiku. We design an inventory reorder function called `check_reorder_status()` that compares current stock against safety thresholds, calculates the replenishment units needed (including a protective safety buffer), and computes the exact capital expenditure required in KES to restock each item before the next major market day.

```python
def check_reorder_status(inventory, safety_buffer_pct=0.50):
    """
    Evaluate stock levels across merchandise categories to identify items requiring wholesale reordering.

    Parameters:
        inventory (dict): Nested dictionary with category keys and properties:
            'current_stock', 'min_threshold', 'unit_cost', 'unit_price', 'weekly_sales'.
        safety_buffer_pct (float, optional): Fractional safety buffer to add above minimum threshold. Defaults to 0.50 (50%).

    Returns:
        dict: Inventory diagnostic containing:
            - reorder_needed (dict): Items needing replenishment with required units and cost.
            - sufficient_stock (dict): Items with safe inventory levels.
            - total_reorder_cost (int/float): Total working capital required in KES.
    """
    reorder_needed = {}
    sufficient = {}
    total_cost = 0

    for category, details in inventory.items():
        stock = details["current_stock"]
        threshold = details["min_threshold"]
        unit_cost = details["unit_cost"]

        if stock < threshold:
            target_level = int(threshold * (1.0 + safety_buffer_pct))
            units_to_order = target_level - stock
            cost = units_to_order * unit_cost
            reorder_needed[category] = {
                "current_stock": stock,
                "min_threshold": threshold,
                "units_to_order": units_to_order,
                "unit_cost": unit_cost,
                "reorder_cost": cost
            }
            total_cost += cost
        else:
            sufficient[category] = {
                "current_stock": stock,
                "min_threshold": threshold,
                "surplus": stock - threshold
            }

    return {
        "reorder_needed": reorder_needed,
        "sufficient_stock": sufficient,
        "total_reorder_cost": total_cost
    }

# Evaluate inventory
reorder_report = check_reorder_status(inventory_data)
print("Inventory Health & Reorder Assessment:")
print("=" * 88)
print(f"{'Product Category':<20} {'Current':>8} {'Min Safe':>9} {'Status':<13} {'Reorder Qty':>12} {'Estimated Cost (KES)':>21}")
print("-" * 88)

for cat, det in inventory_data.items():
    if cat in reorder_report["reorder_needed"]:
        item = reorder_report["reorder_needed"][cat]
        print(f"{cat:<20} {item['current_stock']:>8} {item['min_threshold']:>9} {'REORDER':<13} {item['units_to_order']:>12} {item['reorder_cost']:>21,}")
    else:
        item = reorder_report["sufficient_stock"][cat]
        print(f"{cat:<20} {item['current_stock']:>8} {item['min_threshold']:>9} {'SUFFICIENT':<13} {0:>12} {0:>21,}")

print("-" * 88)
print(f"Total Reorder Capital Required: KES {reorder_report['total_reorder_cost']:,} across {len(reorder_report['reorder_needed'])} depleted categories.")
```

## Task 5: Profit Margin & Operational Efficiency Analysis
**Approach:**  
We analyze the relationship between daily operational costs and revenue. By calculating each day's cost-to-revenue ratio and sorting the days by net profit in descending order, we highlight which days drive the stall's profitability and pinpoint operational drag.

```python
# Analyze cost efficiency and rank days by net profit
ranked_days = []
for item in stall_data:
    rev = item["revenue"]
    exp = item["expenses"]
    profit = rev - exp
    margin = (profit / rev * 100) if rev > 0 else 0.0
    cost_ratio = (exp / rev * 100) if rev > 0 else 0.0
    ranked_days.append({
        "day": item["day"],
        "revenue": rev,
        "expenses": exp,
        "profit": profit,
        "margin": margin,
        "cost_ratio": cost_ratio
    })

# Sort purely with standard Python key function
ranked_days.sort(key=lambda x: x["profit"], reverse=True)

print("Trading Days Ranked by Net Profit Contribution:")
print("-" * 80)
for idx, d in enumerate(ranked_days, start=1):
    print(f"Rank {idx}: {d['day']:<10} | Net Profit: KES {d['profit']:>6,} | Margin: {d['margin']:.1f}% | Cost Ratio: {d['cost_ratio']:.1f}%")

print("-" * 80)
top_2_profit = ranked_days[0]['profit'] + ranked_days[1]['profit']
total_weekly_profit = sum(d['profit'] for d in ranked_days)
pct_share = (top_2_profit / total_weekly_profit) * 100
print(f"Top 2 Days ({ranked_days[0]['day']} + {ranked_days[1]['day']}) generate {pct_share:.1f}% of total weekly net profit!")
```
