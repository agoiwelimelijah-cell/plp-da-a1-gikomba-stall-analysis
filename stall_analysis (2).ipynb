## Data Setup
**Approach:**  
We initialize Wanjiku's 14-day fresh-produce sales records and current inventory stock levels as provided in the assignment dataset.

```python
# Dataset: Wanjiku's Gikomba Market Fresh-Produce Stall (14 Days)
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}

print("Data successfully loaded for Wanjiku's Gikomba produce stall.")
print(f"Total days recorded: {len(daily_sales)}")
print(f"Stock items tracked: {', '.join(stock.keys())}")
```

## Task 1: 14-Day Sales Aggregations & Extremes Identification
**Approach:**  
We calculate the total, average, highest, and lowest daily sales for the full 14 days using standard Python built-ins (`sum`, `len`, `max`, `min`). Using `daily_sales.index()`, we find the index of the highest and lowest amounts to identify the corresponding day name from `days` and determine the week number (positions 0–6 correspond to Week 1, and positions 7–13 correspond to Week 2).

```python
# Task 1: Compute total, average, highest, and lowest sales
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Identify best and worst day positions using list.index()
best_idx = daily_sales.index(highest_sales)
worst_idx = daily_sales.index(lowest_sales)

best_day_name = days[best_idx]
best_week_num = 1 if best_idx < 7 else 2

worst_day_name = days[worst_idx]
worst_week_num = 1 if worst_idx < 7 else 2

print("--- Task 1: Overall 14-Day Sales Metrics ---")
print(f"Total 14-Day Sales:   KES {total_sales:,}")
print(f"Average Daily Sales:  KES {avg_sales:,.2f}")
print(f"Highest Daily Sales:  KES {highest_sales:,} on {best_day_name} (Week {best_week_num})")
print(f"Lowest Daily Sales:   KES {lowest_sales:,} on {worst_day_name} (Week {worst_week_num})")
```

## Task 2: Weekly Totals and Week-on-Week Growth (`week_totals()`)
**Approach:**  
We define the function `week_totals(sales)` that takes the sales list, uses list slicing (`sales[:7]` and `sales[7:]`) to compute Week 1 and Week 2 totals, and returns them as a tuple `(week1_total, week2_total)`. We then unpack the tuple and compute the week-on-week growth percentage: `((week2_total - week1_total) / week1_total) * 100`.

```python
def week_totals(sales):
    """
    Calculate total sales for Week 1 and Week 2 using list slicing.

    Parameters:
        sales (list of int/float): Daily sales for 14 days.

    Returns:
        tuple: A 2-tuple containing (week_1_total, week_2_total).
    """
    week1_total = sum(sales[:7])
    week2_total = sum(sales[7:])
    return (week1_total, week2_total)

# Unpack the returned tuple
week1, week2 = week_totals(daily_sales)
wow_growth = ((week2 - week1) / week1) * 100

print("--- Task 2: Weekly Breakdown & Growth ---")
print(f"Week 1 Total Sales: KES {week1:,}")
print(f"Week 2 Total Sales: KES {week2:,}")
print(f"Week-on-Week Growth: {wow_growth:+.2f}%")
```

## Task 3: Daily Sales Classification (`classify_day()`)
**Approach:**  
We define `classify_day(amount, average)` which categorizes a day as 'Strong' if daily sales exceed 20% above the 14-day average (`amount > 1.20 * average`), 'Weak' if more than 20% below the average (`amount < 0.80 * average`), and 'Normal' otherwise. We then loop across all 14 days and use the dictionary counting pattern (`counts[category] = counts.get(category, 0) + 1`) to tally the occurrences.

```python
def classify_day(amount, average):
    """
    Classify daily sales performance against the overall average.

    Parameters:
        amount (int/float): Sales for the specific day in KES.
        average (float): Mean daily sales across the 14 days in KES.

    Returns:
        str: 'Strong' if > 20% above average, 'Weak' if > 20% below average, else 'Normal'.
    """
    if amount > 1.20 * average:
        return "Strong"
    elif amount < 0.80 * average:
        return "Weak"
    else:
        return "Normal"

# Apply classification to every day and count using dictionary counting pattern
day_counts = {}
print("Day-by-Day Performance Classification:")
print("-" * 70)
for i, amount in enumerate(daily_sales):
    day_name = days[i]
    week_num = 1 if i < 7 else 2
    classification = classify_day(amount, avg_sales)
    day_counts[classification] = day_counts.get(classification, 0) + 1
    print(f"Day {i+1:>2} ({day_name}, W{week_num}): KES {amount:,} -> {classification}")

print("-" * 70)
print(f"Classification Summary: {day_counts}")
```

## Task 4: Weekdays vs Weekends Comparative Analysis
**Approach:**  
In Gikomba Market, foot-traffic and buying behaviors vary significantly between weekdays (Mon–Fri) and weekends (Sat–Sun). We slice and aggregate the weekday sales (days 0–4 and 7–11) and weekend sales (days 5–6 and 12–13), calculate their respective daily averages, and determine the percentage difference.

```python
# Separate weekdays (Mon-Fri) and weekends (Sat-Sun)
weekday_sales = daily_sales[0:5] + daily_sales[7:12]  # 10 weekdays
weekend_sales = daily_sales[5:7] + daily_sales[12:14]  # 4 weekend days

weekday_avg = sum(weekday_sales) / len(weekday_sales)
weekend_avg = sum(weekend_sales) / len(weekend_sales)

# State the difference as a percentage relative to weekday sales
pct_diff = ((weekend_avg - weekday_avg) / weekday_avg) * 100

print("--- Task 4: Weekday vs Weekend Sales Performance ---")
print(f"Weekday Average (Mon-Fri, 10 days): KES {weekday_avg:,.2f}")
print(f"Weekend Average (Sat-Sun, 4 days):   KES {weekend_avg:,.2f}")
print(f"Absolute Difference:                KES {weekend_avg - weekday_avg:,.2f}")
print(f"Weekend sales are {pct_diff:+.2f}% higher than weekday sales!")
```

## Task 5: Inventory Reorder Evaluation & Gross Margin Analysis
**Approach:**  
We iterate through the `stock` dictionary using `.items()` to:  
1. Identify items where quantity in stock is at or below the reorder level (`qty <= reorder`).  
2. Calculate the gross margin percentage for each product using the formula: `(price - cost) / price * 100`.

```python
def check_reorder_and_margins(stock_dict):
    """
    Identify products at or below reorder levels and calculate gross margins.

    Parameters:
        stock_dict (dict): Nested dictionary with keys 'qty', 'reorder', 'cost', and 'price'.

    Returns:
        tuple: (list of products needing reorder, dict of gross margins by product).
    """
    reorder_list = []
    margins = {}

    for product, info in stock_dict.items():
        # Check if qty is at or below reorder level
        if info["qty"] <= info["reorder"]:
            reorder_list.append(product)
        
        # Gross margin percentage = (price - cost) / price * 100
        margin_pct = ((info["price"] - info["cost"]) / info["price"]) * 100
        margins[product] = margin_pct

    return (reorder_list, margins)

reorder_items, gross_margins = check_reorder_and_margins(stock)

print("--- Task 5: Inventory Health & Gross Margin Evaluation ---")
print(f"{'Product':<15} {'Qty (kg)':>9}  {'Reorder Level':>13}  {'Gross Margin (%)':>16}  {'Status':<14}")
print("-" * 70)
for product, info in stock.items():
    status = "REORDER NEEDED" if product in reorder_items else "Stock Safe"
    print(f"{product:<15} {info['qty']:>9}  {info['reorder']:>13}  {gross_margins[product]:>15.2f}%  {status:<14}")

print("-" * 70)
print(f"Products needing urgent reorder: {reorder_items}")
best_margin_item = max(gross_margins.items(), key=lambda x: x[1])
print(f"Highest gross margin product: {best_margin_item[0]} ({best_margin_item[1]:.2f}% margin)")
```

## Task 6: Printed Stall Performance Report with KES Formatting
**Approach:**  
We construct a clean, aligned, and professional ASCII 'Stall Performance Report' utilizing Python f-strings, fixed-width character padding, and comma thousands separators for all monetary values in KES.

```python
def print_stall_performance_report():
    """
    Print a clean, aligned, and professional Stall Performance Report for Wanjiku.
    All currency values formatted with comma thousands separators in KES.
    """
    print("=" * 72)
    print("                WANJIKU'S GIKOMBA PRODUCE STALL                         ")
    print("                   STALL PERFORMANCE REPORT                             ")
    print("=" * 72)
    print(f"{'Day':<5} {'Weekday':<9} {'Week':<6} {'Daily Sales (KES)':>17}   {'Performance Classification':<28}")
    print("-" * 72)
    
    for i, amount in enumerate(daily_sales):
        day_name = days[i]
        week_num = 1 if i < 7 else 2
        cls = classify_day(amount, avg_sales)
        detail = "Strong (>20% of avg)" if cls == "Strong" else ("Weak (<20% of avg)" if cls == "Weak" else "Normal")
        print(f"{i+1:>2}    {day_name:<9} {week_num:<6} {amount:>17,}   {detail:<28}")
        
    print("-" * 72)
    print("AGGREGATE SUMMARY:")
    print(f"  * Total 14-Day Sales:       KES {total_sales:,}")
    print(f"  * Daily Average Sales:      KES {avg_sales:,.2f}")
    print(f"  * Week 1 Total:             KES {week1:,}")
    print(f"  * Week 2 Total:             KES {week2:,}")
    print(f"  * Week-on-Week Growth:      {wow_growth:+.2f}%")
    print(f"  * Highest Day:              KES {highest_sales:,} on {best_day_name} (Week {best_week_num})")
    print(f"  * Lowest Day:               KES {lowest_sales:,} on {worst_day_name} (Week {worst_week_num})")
    print(f"  * Weekday Average (10 days): KES {weekday_avg:,.2f}")
    print(f"  * Weekend Average (4 days):  KES {weekend_avg:,.2f} ({pct_diff:+.2f}% vs weekdays)")
    print("=" * 72)
    print("INVENTORY & REORDER ALERT:")
    print(f"  * Products to Restock Now:  {', '.join([f'{p} ({stock[p]["qty"]}kg <= {stock[p]["reorder"]}kg)' for p in reorder_items])}")
    print(f"  * Top Gross Margin Product:  Avocado ({gross_margins['Avocado']:.2f}% margin), Sukuma ({gross_margins['Sukuma']:.2f}% margin)")
    print("=" * 72)

# Run and print the report
print_stall_performance_report()
```
